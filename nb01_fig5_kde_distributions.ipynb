# Fig 5 - KDE distributions of human ratings vs. LLM expected scores
Smooth-density comparison of human interoceptivity ratings against the three LLMs' expected scores.

## Load data

```python
"""Load the merged data using the project's data_loading module."""
from data_loading import load_data

df = load_data()
df.head()
```

## Build the figure

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

series = [
    ("Human ratings",          df["human"].values,                   "#1f77b4"),
    ("Qwen 2.5-72B Instruct",  df["qwen_expected"].values,           "#9467bd"),
    ("Llama 3.3-70B Instruct", df["llama_instruct_expected"].values, "#2ca02c"),
    ("Llama 3.1-70B Base",     df["llama_base_expected"].values,     "#d62728"),
]

x_grid = np.linspace(0, 5, 600)
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

for label, values, color in series:
    kde = gaussian_kde(values, bw_method=0.25)
    density = kde(x_grid)
    ax.plot(x_grid, density, label=label, color=color, linewidth=2.2)
    ax.fill_between(x_grid, density, alpha=0.10, color=color)
    ax.axvline(np.mean(values), color=color, linestyle=":", linewidth=1.0, alpha=0.55)

ax.set_xlim(0, 5)
ax.set_ylim(bottom=0)
ax.set_xlabel("Interoceptivity score (0-5)", fontsize=12)
ax.set_ylabel("Density", fontsize=12)
ax.set_title("Distribution of interoceptivity scores: human ratings vs. LLM expected scores",
             fontsize=13, fontweight="bold")
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(frameon=True, fontsize=11, loc="upper right")
plt.tight_layout()
plt.savefig("fig5_kde_distributions.png", dpi=200, bbox_inches="tight")
plt.show()
```

## Summary stats

```python
for label, values, _ in series:
    print(f"{label:25s}  mean={np.mean(values):.3f}  "
          f"median={np.median(values):.3f}  sd={np.std(values):.3f}")
```
